# 1. Simulasi undo sederhana

```python
# Fungsi Operasi Stack
def stack():
    s = []
    return (s)
def push(s,data):
    s.append(data)
def pop(s):
    data = s.pop()
    return (data)
def peek(s):
    return (s[len(s)-1])
def isEmpty(s):
    return (s==[])
def size(s):
    return (len(s))

# Main Fungsi
def editTeksUndo():
    st = stack()
    stop = False

    while not stop:
        teks = input("Input :")
        if teks != "undo":
            push(st,teks)
        else:
            pop(st)
            stop = True
    return print('Output:',st)

# Main Program
editTeksUndo()
```

# 2. Menghitung banyak pasangan tanda kurung valid

```python
# Main Fungsi
def parenthesesCheck(strMath):
    s = stack()
    matched = True

    kurung = {")":"(","}":"{","]":"["}
    kurungBuka = kurung.values()
    kurungTutup = kurung.keys()

    for ch in strMath:
        if ch in kurungBuka:
            push(s,ch)
        elif ch in kurungTutup:
            if isEmpty(s):
                matched = False
            else:
                temp = pop(s)
        
                if kurung[ch] != temp:
                    matched = False

    if not isEmpty(s):
        matched = False
    
    if matched:
      return "Seimbang"
    else:
      return "Tidak Seimbang"  
    


# Main Program
mathStr1 = '((A+B) *(C-D))'
print(parenthesesCheck(mathStr1))

print("="*10)
mathStr2 = "((A+B) *(C-D)"
print(parenthesesCheck(mathStr2))
```

# 3. Selection sort dan posisi minimum

```python
# Main Fungsi
def selectionSort(listData):
    print('Algoritma Selection Sort konvensional')
    print('Data Awal=',listData)
    for outIter in range(len(listData)-1):
        minIndex=outIter
        for i in range(outIter+1,len(listData)):
            if listData[i]<listData[minIndex]:
                minIndex=i
                minimum = listData[minIndex]

        print("Indeks minimum:",minIndex,"\ndengan nilai minimum:",minimum)
        temp=listData[outIter]
        listData[outIter]=listData[minIndex]
        listData[minIndex]=temp
        print('Iterasi ke-',outIter+1,':',listData,"\n")
    print('Data Urut=',listData)
# Main Program
data = [23, 78, 45, 8, 32, 56, 1]
selectionSort(data)
```
