# **BINARY SEARCH TREE**

## **Konsep Dasar Binary Search Tree**

Binary Search Tree (BST) merupakan jenis struktur data Binary Tree khusus dimana properti berikut harus dipenuhi untuk setiap node "X" di dalam tree:

1. Left child (anak kiri) dari node X dan seluruh keturunannya memiliki nilai data yang lebih kecil dari nilai X.

2. Right child (anak kanan) dan seluruh keturunannya memiliki nilai data yang lebih besar dari nilai X.

3. Subtree kiri maupun subtree kanan dari sebuah node juga harus berupa sebuah Binary Search Tree.

Keuntungan utama dari BST adalah membuat operasi seperti mencari (search), menambahkan (insert), dan menghapus (delete) nilai menjadi jauh lebih cepat dibandingkan binary tree biasa, serta dilakukan tanpa perlu melakukan pergeseran elemen memori seperti pada tipe data Array. Agar implementasi berjalan dengan optimal, diasumsikan bahwa semua nilai pada sebuah Binary Search Tree bersifat unik (tidak ada duplikat).

## **Traversal pada Biary Search Tree**

Untuk mengecek apakah sebuah Binary Tree sudah valid menjadi BST, kita dapat melakukan In-Order Traversal (penelusuran berurutan). Jika tree tersebut benar adalah BST, hasil traversal pasti akan menghasilkan urutan data yang terus meningkat dari kecil ke besar.

### Code

```python
class TreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

def inOrderTraversal(node):
    if node is None:
        return
    inOrderTraversal(node.left)
    print(node.data, end=", ")
    inOrderTraversal(node.right)

root = TreeNode(13)
node7 = TreeNode(7)
node15 = TreeNode(15)
node3 = TreeNode(3)
node8 = TreeNode(8)
node14 = TreeNode(14)
node19 = TreeNode(19)
node18 = TreeNode(18)

root.left = node7
root.right = node15

node7.left = node3
node7.right = node8

node15.left = node14
node15.right = node19

node19.left = node18

# Traverse
inOrderTraversal(root)
```

## 1. Operasi Pencarian (Search)
Pencarian data pada BST dilakukan dengan sangat efisien karena kita tidak perlu mengecek setiap node. Mulai dari root, kita dapat memeriksa:

- Apakah nilai sama dengan target? Jika iya, node telah ditemukan.

- Apakah target lebih kecil? Berarti kita hanya perlu mencari di subtree kiri.

- Apakah target lebih besar? Berarti kita hanya perlu mencari di subtree kanan.

### Code

```python
def search(node, target):
    if node is None:
        return None 
    elif node.data == target:
        return node
    elif target < node.data:
        return search(node.left, target)
    else:
        return search(node.right, target)
```

## 2. Menyisipkan Node Baru (Insert)
Proses penyisipan node (insert) memiliki logika yang serupa dengan pencarian.

1. Mulai dari root node.

2. Jika node tujuan kosong (None), buat dan kembalikan node baru disana (node akan menjadi node daun/leaf baru).

3. Jika nilai baru lebih kecil, turun ke anak kiri; jika lebih besar, turun ke anak kanan.

### Code

```python
def insert(node, data):
    if node is None:
        return TreeNode(data)
    else:
        if data < node.data:
            node.left = insert(node.left, data)
        elif data > node.data:
            node.right = insert(node.right, data)
    return node
```

### 3. Menghapus Node (Delete)
Penghapusan node pada BST sedikit lebih kompleks, karena setelah sebuah node dihapus, koneksi pada struktur tree harus tetap terjaga (tidak boleh terputus). Terdapat 3 skenario/kasus saat menghapus node:

- Kasus 1: Node yang dihapus adalah daun (Leaf Node) yang tidak memiliki anak. Node bisa langsung dihapus/diputus.

- Kasus 2: Node memiliki satu child. Kita langsung menyambungkan 'parent' dari node yang dihapus dengan anak satu-satunya tersebut.

- Kasus 3: Node memiliki dua buah children. Kita harus mencari in-order successor (node dengan nilai terkecil pada subtree kanan), menukar nilainya dengan node yang ingin dihapus, kemudian menghapus successor aslinya tersebut dari tree.

Oleh karena itu, sebelum membuat fungsi delete, kita membutuhkan bantuan fungsi pencarian nilai terkecil minValueNode().

### Code

```python
def minValueNode(node):
    current = node
    while current.left is not None:
        current = current.left
    return current
```

```python
def delete(node, data):
    if not node:
        return None

    if data < node.data:
        node.left = delete(node.left, data)
    elif data > node.data:
        node.right = delete(node.right, data)
    else:
        # Node with only one child or no child
        if not node.left:
            temp = node.right
            node = None
            return temp
        elif not node.right:
            temp = node.left
            node = None
            return temp

        # Node with two children, get the in-order successor
        node.data = minValueNode(node.right).data
        node.right = delete(node.right, node.data)

    return node
```

## **Kesimpulan Waktu Eksekusi (Time Complexity)**
Mencari nilai pada BST sama cepatnya dengan mencari nilai pada Array yang telah disorting, yaitu dengan kompleksitas waktu rata-rata sebesar $O(\log n)$ atau $O(h)$ dimana h adalah tinggi dari tree. Pada skenario terburuk, jika bentuk tree sangat panjang menjuntai ke bawah seperti garis lurus (disebut juga unbalanced tree), performanya akan mirip dengan Linked List yaitu $O(n)$.
