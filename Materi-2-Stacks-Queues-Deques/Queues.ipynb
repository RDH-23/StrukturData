## **Definisi Queues**

**Queues** atau antrian merupakan struktur data dimana penambahan data baru dan penghapusan data berada di ujung yang berbeda. Hal ini berbeda dengan stacks, dimana penambahan data baru dan penghapusan data, dilakukan pada ujung yang sama.<br>
Pada **Queues**, seperti halnya antrian, penambahan data bari dilakukan di suatu ujung atau yang dikenal dengan **rear**, dan penghapusan data dilakukan pada ujung yang dikenal dengan **front**, seperti yang terlihat pada gambar 1 berikut:

**Gambar 1. Queues atau Antrian**

Jika konsep penambahan atau penghapusan data pada stacks dikenal dengan nama LIFO (Last In First Out), maka pada Queues menggunakan konsep **FIFO (First In FIrst Out)**

```python
def createQueue():
    q = []
    return q
def enqueue(q,data): #front - indeks 0
    q.insert(0,data)
    return (q)
def dequeue(q): #rear - indeks terakhir
    data = q.pop()
    return (data)
def isEmpty(q):
    return (q==[])
def size(q):
    return (len(q))
```

## **Implementasi-Queues**

Contoh implementasi queues ini adalah permainan anak-anak, yaitu ular naga. Pada permainan ini, sekelompok anak berbaris dan terdapat dua buah anak sebagai penjaga, yang bertugas untuk 'menangkap anak' seiring dengan 'ular naga' berakhir. Permainan akan berakhir jika hanya eterdapat satu anak yang memenagkan permainan, yaitu anak tersebut lolo dari tangkapan penjaga, seperti yang terlihat pada Gambar 4

**Gambar 4. Permainan Ular Naga**

Berikut code simulasi dari permainan tersebut, dengan sedikit modifikasi, yaitu fungsi penjaga dan lagu 'ular naga' digantikan dengan bilangan tertentu, dan sekelompok anak disusun dalam suatu antrian. Permainan akan terus berjalan, sampai dengan hitungan bilangan tertentu tersebut, Jika pada hitungan tertentu, terdapat anak yang berada pada ujung antrian (*front*), maka anak tersebut haru keluar barisan antrian. Permainan dimenangkan oleh seorang anak yang tersisa di dalam antrian.

### Code

```python
def ularNaga(nama,hitungan):
    gameQueue = createQueue()
    for namaAnak in nama:
        enqueue(gameQueue,namaAnak)
    print('Peserta Permainan=',gameQueue)
    while size(gameQueue) > 1:
        for i in range(hitungan):
            enqueue(gameQueue,dequeue(gameQueue))
            print('hitungan ke-',i,'=',gameQueue)
        dequeue(gameQueue)
        print('Peserta Permainan=',gameQueue)
    return dequeue(gameQueue)
```

```python
ularNaga(['andi','rita','sari','anton','rafa','diana','zaki'],2)
```

# **Scheduling**

1. Pendahuluan: Konsep Resource Sharing
   - Resource Sharing (berbagi sumber daya) terjadi ketika beberapa proses atau tugas (tasks) mencoba mengakses satu sumber daya yang sama secara bersamaan.
   - Contoh utama: Beberapa aplikasi (seperti Anaconda, Microsoft Office) berjalan bersamaan dan membutuhkan Processor (CPU).
   - Masalah: Prosesor tidak dapat menangani semua tugas sekaligus pada waktu yang sama.
   - Solusi: Prosesor menggunakan Penjadwalan (Scheduling) untuk mengatur antrian tugas.

2. Mekanisme Penjadwalan (Time Limiting)
- Prosesor memberikan batas waktu tertentu untuk setiap tugas, disebut CPU Time atau Time Quantum.
- Dalam video ini, CPU Time ditetapkan selama 3 detik.
- Aturan Antrian (Queue Rules):
  1. Tugas diproses sesuai waktu yang ditentukan (maksimal 3 detik).
  2. Jika tugas selesai kurang dari 3 detik, tugas keluar dari antrian.
  3. Jika tugas membutuhkan waktu lebih dari 3 detik, sisa waktu tugas tersebut akan dimasukkan kembali ke antrian belakang.

3. Ilustrasi Studi Kasus (Simulasi)
- Tugas (Tasks):
- Task A: Butuh 5 detik 
- Task B: Butuh 4 detik 
- Task C: Butuh 2 detik 
- Kapasitas CPU: 3 detik
Langkah-langkah Simulasi:
1. Iterasi 1 : CPU memproses A (3 detik). Sisa A = 2 detik. Antrian: B, C, A(sisa).
2. Iterasi 2 : CPU memproses B (3 detik). Sisa B = 1 detik. Antrian: C, A(sisa), B(sisa).
3. Iterasi 3 : CPU memproses C (2 detik). C selesai. Antrian: A(sisa), B(sisa).
4. Iterasi 4 : CPU memproses A (sisa 2 detik). A selesai. Antrian: B(sisa).
5. Iterasi 5 : CPU memproses B (sisa 1 detik). B selesai. Antrian kosong.

4. Hasil Akhir (Total Waktu per Tugas)
Berdasarkan simulasi, total waktu yang dibutuhkan setiap tugas dari mulai antri hingga selesai adalah:

- Task A: 10 satuan waktu
- Task B: 11 satuan waktu
- Task C: 8 satuan waktu

## **Code**

```python
def inputTask(numOfTask):
    task={}
    for i in range(numOfTask):
        nama=input("Nama Task = : ")
        waktu=int(input("Waktu Task = "))
        task[nama]=[waktu,0]
    return task
```

```python
Task=inputTask(3)
```

```python
print(Task)
```

```python
def scheduling(limitTime,task):
    taskQueue=createQueue()
    for nama in task.keys():
        enqueue(taskQueue,nama)
    total=0
    while not(isEmpty(taskQueue)):
        temp=dequeue(taskQueue)
        remainTime=task[temp][0]-limitTime

        if remainTime>0:
            enqueue(taskQueue,temp)
            procTime=limitTime
        else:
            procTime=task[temp][0]
            remainTime=0

        total=total+procTime
        task[temp][0]=remainTime
        task[temp][1]=total
        
        print(taskQueue)
    return task
```

**Soal 1: Pemisahan Angka Ganjil dan Genap**
Buatlah sebuah fungsi bernama pisahkan_ganjil_genap(daftar_angka). Fungsi ini menerima sebuah list berisi angka acak.

- Tugas: Buat dua queue baru di dalam fungsi tersebut (misal: q_ganjil dan q_genap). Masukkan angka ganjil ke q_ganjil dan angka genap ke q_genap menggunakan enqueue().

- Output: Keluarkan (dequeue) semua isi q_genap terlebih dahulu dan cetak ke layar, kemudian disusul mengeluarkan semua isi q_ganjil.

```python
def pisahkan_ganjil_genap(daftar_angka):
    # Membuat dua queue terpisah
    q_ganjil = createQueue()
    q_genap = createQueue()
    
    # Memisahkan angka
    for angka in daftar_angka:
        if angka % 2 == 0:
            enqueue(q_genap, angka)
        else:
            enqueue(q_ganjil, angka)
            
    print("--- Mengeluarkan Angka Genap ---")
    while not isEmpty(q_genap):
        print(dequeue(q_genap))
        
    print("--- Mengeluarkan Angka Ganjil ---")
    while not isEmpty(q_ganjil):
        print(dequeue(q_ganjil))

# Main Program
angka_acak = [1, 4, 5, 8, 9, 10, 15, 22]
pisahkan_ganjil_genap(angka_acak)
```

**Soal 2: Sistem Antrean Drive-Thru Sederhana**
Buatlah dua buah fungsi terpisah untuk menyimulasikan kasir drive-thru:

1. tambah_pesanan(q, pesanan): Memasukkan pesanan pelanggan ke dalam queue dan mencetak "Pesanan [pesanan] masuk antrean".

2. layani_pesanan(q): Mengecek apakah queue kosong. Jika tidak kosong, keluarkan pesanan menggunakan dequeue() dan cetak "Sedang menyiapkan: [pesanan]". Jika kosong, cetak "Antrean kosong, tidak ada pesanan".

```python
def tambah_pesanan(q, pesanan):
    enqueue(q, pesanan)
    print(f"Masuk antrean: {pesanan}")

def layani_pesanan(q):
    if isEmpty(q):
        print("Antrean kosong, tidak ada pesanan.")
    else:
        pesanan_sekarang = dequeue(q)
        print(f"Sedang menyiapkan: {pesanan_sekarang}")

# --- Contoh Penggunaan ---
antrean_drivethru = createQueue()

print("--- Jam Sibuk Dimulai ---")
tambah_pesanan(antrean_drivethru, "Burger & Kentang")
tambah_pesanan(antrean_drivethru, "Es Kopi Susu")
tambah_pesanan(antrean_drivethru, "Ayam Goreng")

print(f"Total pesanan menunggu: {size(antrean_drivethru)}\n")

print("--- Mulai Melayani ---")
layani_pesanan(antrean_drivethru) # Melayani Burger
layani_pesanan(antrean_drivethru) # Melayani Es Kopi
layani_pesanan(antrean_drivethru) # Melayani Ayam Goreng

# Mencoba melayani saat antrean sudah habis
layani_pesanan(antrean_drivethru)
```

**Soal 3: Rotasi Antrean (Shift Queue)**
Terkadang dalam penjadwalan, antrean yang berada di depan dipindahkan ke posisi paling belakang agar yang lain bisa maju.
Buatlah fungsi rotasi_antrean(q, n). Fungsi ini menerima sebuah queue yang sudah ada isinya, dan angka n.

- Tugas: Pindahkan elemen yang berada di paling depan (front) ke posisi paling belakang (rear) sebanyak n kali. Gunakan kombinasi dequeue() dan enqueue().

```python
def rotasi_antrean(q, n):
    print(f"Antrean awal: {q}")
    print(f"Melakukan rotasi sebanyak {n} kali...\n")
    
    for i in range(n):
        # 1. Ambil data paling depan (dequeue)
        data_depan = dequeue(q)
        
        # 2. Masukkan kembali ke paling belakang (enqueue)
        enqueue(q, data_depan)
        
        print(f"Rotasi ke-{i+1}: {q}")
        
    return q

# --- Contoh Penggunaan ---
antrean_tugas = createQueue()
enqueue(antrean_tugas, "Tugas A")
enqueue(antrean_tugas, "Tugas B")
enqueue(antrean_tugas, "Tugas C")
enqueue(antrean_tugas, "Tugas D")

# Menggeser antrean 2 kali
rotasi_antrean(antrean_tugas, 2)
```

```python
Pseudocode enqueue:
Procedure Enqueue(queue, data):
    IF queue is full THEN
        PRINT "Queue Overflow"
    ELSE
        Tambahkan data ke posisi paling belakang (REAR)
    ENDIF
End Procedure

Pseudocode dequeue:
Procedure Dequeue(queue):
    IF queue is empty THEN
        PRINT "Queue Underflow"
    ELSE
        Ambil data dari posisi paling depan (FRONT)
        Geser semua elemen setelahnya ke depan
        Return data tersebut
    ENDIF
End Procedure
```

**9.** Cek palindrome menggunakan deque

```python
def createQueue():
    return []

def enqueue(q, data):
    q.append(data)
    return q

def dequeue(q):
    return q.pop(0)

def isEmpty(q):
    return q == []

def size_manual(q):
    # Menghitung size tanpa len() dengan bantuan queue sementara
    temp_q = createQueue()
    count = 0
    while not isEmpty(q):
        enqueue(temp_q, dequeue(q))
        count += 1
    # Kembalikan data ke queue asli
    while not isEmpty(temp_q):
        enqueue(q, dequeue(temp_q))
    return count

def cekPalindrom(teks):
    teks = teks.lower().replace(" ", "")
    q = createQueue()
    
    # Masukkan semua karakter ke queue
    for char in teks:
        enqueue(q, char)
    
    status = True
    
    # Selama karakter di queue masih lebih dari 1
    while size_manual(q) > 1:
        # 1. Ambil karakter depan (FRONT)
        depan = dequeue(q)
        
        # 2. Cari karakter belakang (REAR)
        # Rotasi queue sebanyak (size - 1) kali
        s = size_manual(q)
        for i in range(s - 1):
            enqueue(q, dequeue(q))
        
        # Sekarang karakter terakhir sudah berada di posisi paling depan
        belakang = dequeue(q)
        
        # 3. Bandingkan
        if depan != belakang:
            status = False
            break
            
    return status

# --- Uji Coba ---
kata = "Radar"
if cekPalindrom(kata):
    print(f"'{kata}' adalah Palindrom")
else:
    print(f"'{kata}' bukan Palindrom")
```
