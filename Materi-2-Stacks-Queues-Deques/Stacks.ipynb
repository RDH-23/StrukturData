## **Definisi Stacks**

**Stacks** merupakan salah satu struktur data dimana penambahan dan penghapusan data, hanya dapat dilakukan pada satu ujung yang **sama**, atau yang bisa dikenal dengan istilah **top**.

**Stacks (Tumpukan)** adalah sebuah tipe data abstrak linier yang berfungsi sebagai kumpulan elemen, di mana penambahan elemen baru dan penghapusan elemen yang sudah ada selalu terjadi di tempat yang sama.

Semakin data jauh berada dari posisi top, maka data tersebut diindikasikan berda di stack lebih lama dibandingkan dengan data yang berada dekat pada posisi top.

Jika terdapat data baru yang ditambahkan di stack, maka data ini pulalah yang akan dihapus ketika terdapat proses penghapusan data. Konsep ini dikenal denagan anam **LIFO-last In First Out**.

### **Prinsip LIFO (Last In First Out)**

Prinsip LIFO berarti elemen yang paling akhir ditambahkan ke dalam stack adalah elemen pertama yang akan dihapus atau diambil.
* Elemen baru selalu ditambahkan (didorong/*pushed*) ke posisi paling atas.
* Penghapusan (*pop*) juga hanya terjadi dari posisi atas.
* Aturan ini memastikan urutan yang ketat : **terakhir masuk -> pertama keluar**

## **Operasi pada Stacks**

Operasi dasar pada stacks

* **stack()**, inisialisasi stack yang kosong
* **push(data)**, penambahan data baru pada posisi top dari stack
* **pop()**, penghapusan data yang terdapat di posisi top dari stack. return value dari fungsi ini adalah data yang dihapus dari stack tersebut
* **peek()**, informasi data yang terletak pada posisi top
* **isEmpty()**, untuk memeriksa apakah stack dalam keadaan kosong
* **size()**, informasi jumlah data yang terdapat pada stack

Buatlah fungsi untuk mengecek apakah suatu string merupakan palindrome menggunakan stack.<br>
Input:"KATAK"<br>
Output:Palindrome

```python
def isPalindrome(kata):
    # Menampung string awal
    s = stack()
    for ch in kata:
        push(s,ch)

    # Menampung string setelah diambil dari belakang
    temp = ''
    while not(isEmpty(s)):
        temp = temp+pop(s)

    if temp==kata:
        return "Palindrome"
    else:    
        return "Bukan palindrome"
```

```python
def hapusBagianTengah(s,removeAngka):
    st = stack()

    while not(isEmpty(s)):
        angka = pop(s)
        if angka != removeAngka:
            push(st,angka)

    while not(isEmpty(st)):
        push(s,pop(st))
    return s
```

## **Delimiter Matching**

**Parentheses** sering digunakan untuk urutan penyelesaian dalam persamaan Matematika, seperti contoh berikut:

P = 5 x (4 + 5)/(3 + 2)x(10-8)

Dengan penggunaan **parentheses** maka persamaan tersebut dapat diselesaikan secara benar, hanya saja terkadang **parentheses** yang ditulis tidak lengkap, sehingga hasil persamaan tidak dapat dikerjakan atau tidak sesuai dengan yang diinginkan, seperti contoh berikut:

P = 5 x (4 + 5)/(3 + 2)x(10-8)

Oleh karena itu diperlukan algoritma untuk pemeriksaan parentheses ini atau secara umum **delimiter**, yaitu '{}','[]', dan '()'.

Beberapa tahapan atau algoritma untuk pengecekan *parentheses* ini adalah:<br>
1. Baca setiap karakter yang terdapat pada suatu string matematika
2. Jika karakter adalah kurung buka, maka masukkan dalam stack
3. Jika karakter adalah tutup, ada beberapa kondisi yang harus diperika, yaitu:
   * Jika stack dalam keadaan empty, maka jumlah kurng tutup lebih banyak daripada kurung siku
   * Jika stack dalam keadaan tidak empty, maka pop stack, dan cocokkan karakter hasil pop dengan kurung tutup, jika sejenis, maka *matched*, jika tidak maka jenis kurung tidak sama
4. Jika semua karakter telah terbaca, dan stack masih dalam keadaan terisi (tidak empty), maka jumlah kurung buka lebih banyak daripada kururng tutup

```python
def checkParentheses(strMat):
    data = stack()
    for ch in strMat:
        if ch == '(':
            push(data,ch)
        elif ch == ')':
            if isEmpty(data):
                return 'kelebihan kurung tutup'
            else:
                pop(data)
    if isEmpty(data):
        return True
    else:
        return 'Kelebihan kurung buka'
```

Berikut adalah fungsi untuk pengecekan kurung pada suatu string matematika

```python
def stack():
    s = []
    return (s)
def push(s,data):
    s.append(data)
def pop(s):
    data = s.pop()
    return (data)
def peek(s):
    return (s[len(s)-1])
def isEmpty(s):
    return (s==[])
def size(s):
    return (len(s))
```

```python
def parenthesesCheck(strMath):
    kurung = {')':'(','}':'{',']':'['}
    kurungBuka=kurung.values()
    kurungTutup=kurung.keys()
    temp=stack()
    matched=True
    for ch in strMath:
        if ch in kurungBuka:
            push(temp,ch)
        if ch in kurungTutup:
            if isEmpty(temp):
                return 'Kelebihan kurung Tutup'
            else:
                tempKurung=pop(temp)
                if tempKurung==kurung[ch]:
                    matched=matched and True
                else:
                    matched=matched and False
    if not(isEmpty(temp)):
        return 'Kelebihan Kurung Buka'
    if matched == True:
        return 'OK'
    else:
        return 'Tidak Cocok Kurungnya'
```

```python
parenthesesCheck('{5+2}*3')
```

```python
def convertBinary(num):
    temp=stack()
    divNum=num
    while divNum>0:
        push(temp,divNum%2)
        divNum=divNum//2
    tempStr=''
    for i in range(size(temp)):
        tempStr=tempStr+str(pop(temp))
    return tempStr

print(convertBinary(25))
```

```python
def paranthesesCheck(strMath):
    operandStack=stack()
    lenMath=len(strMath)
    openOperand='({['
    closeOperand=')}]'
    i = 0
    Matched = True
    while i<(lenMath):
        if strMath[i] in openOperand:
            push(operandStack,strMath[i])
        elif strMath[i] in closeOperand:
            if not (isEmpty(operandStack)):
                top = pop(operandStack)
                if openOperand.index(top)==closeOperand.index(strMath[i]):
                    Matched=Matched and True
                else:
                    Matched=Matched and False
                    print('Kurung Buka dan Kurung Tutup tidak Cocok')
            else:
                Matched=Matched and False
                print('Jumlah Kurung Tutup lebih banyak')
        i=i+1

    if not(isEmpty(operandStack)):
        Matched=False
        print('Jumlah Kurung Buka Lebih banyak')
    return(Matched)
```

## **Ekpresi Aritmatik Infix, Prefix, Postfix**

Ekpresi aritmatik yang sering diteui adalah operasi aritmatik **infix**. Pada operasi aritmatik **infix** ini, **operator** (+,-,*,/,dll) berada diantara dua buah **operand** (bilangan). Ekspresi aritmatik ini dapat dengan mudah diselesaikan ketika tidak ada ambiguitas, seperti:

$$A + B$$
$$A \ast X$$
$$A - B + C$$

Akan tetapi, ekspressi aritmatik ini akan sulit diselesaikan jika terjadi ambiguitas, seperti ekspressi
aritmatik berikut ini:


$$A + B \ast C$$

Pada ekspressi tersebut, terdapat ambiguitas, yaitu apakah A + B diselesaikan terlebih dahulu
ataukah B * C yang harus diselesaikan terlebih dahulu. Ambiguitas ini dapat diatasi dengan
konsep **operator precedence**. Pada operator precedence, operator-operator dibagi menjadi
beberapa level berdasarkan prioritas urutan penyelesaian.


Pada contoh ekspressi aritmatik A + B * C, berdasarkan operator precedence, maka urutan
penyelesaian adalah B * C kemudian hasilnya ditambahkan dengan A. Permasalahan terjadi
ketika pada ekspressi aritmatik A + B * C, A + B ingin diselesaikan terlebih dahulu. Untuk
penyelesaian ekspressi aritmatik yang tidak sesuai dengan urutan operator precedence, maka
tanda '( )' sangatlah diperlukan, sehingga operasi tersebut menjadi (A + B) * C.


Tanda '( )' **tidaklah diperlukan** pada ekspressi aritmatik **prefix** dan **postfix**. Pada ekspressi
aritmatik **prefix**, operator mendahului operand, sedangkankan ekspressi aritmatik **postfix**,
operator ditulis setelah operand.


Berikut contoh ekspresi aritmatik **infix** dan konversinya ke **prefix** dan **postfix**.

$$
\begin{array}{ccc}
\mathbf{Infix} & \mathbf{Prefix} & \mathbf{Postfix} \\
\hline
A+B & +AB & AB+ \\
(A+B)*C & *+ABC & AB+C* \\
(A+B)*(C-D) & *+AB-CD & AB+CD-* \\
A+B*C-D & -+A*BCD & ABC*+D- \\
A+B-C+D & +-+ABCD & AB+C-D+ \\
A*B-C*D & -*AB*CD & AB*CD*-
\end{array}
$$

### **Konversi Infix ke Postfix**

Berikut algoritma untuk konversi ekspressi aritmatik infix ke postfix :
1. Buat struktur data stack untuk menampung operator
2. Baca ekspressi aritmatik dari kiri ke kanan tiap token :
* jika token yang dibaca adalah operand maka masukkan operand tersebut ke dalam
output string
* jika token yang dibaca adalah kurung buka maka push kurung buka tersebut ke dalam
stack
* jika token yang dibaca adalah kurang tutup maka pop stack semua token sampai
ditemukan kurung buka
* jika token yang dibaca adalah operator maka :
    - pop operator-operator yang memiliki precedence lebih tinggi atau sama dan
masukkan operator tersebut ke dalam output string
    - push token operator ke dalam stack
3. Jika masih terdapat operator pada stack, maka pop operator yang tersisa dan letakkan pada
output string


Contoh konversi dari infix ke postfix dapat dilihat pada Gambar 5.


### Infix to Postfix

#### Code 1

```python
def infix_ke_postfix(ekspresi):
    precedence = {'+': 1, '-': 1, '*': 2, '/': 2, '^': 3}
    
    s = stack() 
    output = [] 
    
    ekspresi = ekspresi.replace(" ", "")
    
    for token in ekspresi:
        if token.isalnum():
            output.append(token)
        elif token == '(':
            push(s, token) # Menggunakan fungsi push()
        elif token == ')':
            # Pop stack semua token sampai ditemukan kurung buka
            while not isEmpty(s) and peek(s) != '(':
                output.append(pop(s)) 
            if not isEmpty(s) and peek(s) == '(':
                pop(s)
                
        # Jika token adalah operator
        else:
            # Pop operator yang memiliki precedence lebih tinggi atau sama
            while (not isEmpty(s) and peek(s) != '(' and 
                   precedence.get(peek(s), 0) >= precedence.get(token, 0)):
                output.append(pop(s))
            
            # Push token operator ke dalam stack
            push(s, token)
            
    # 3. Jika masih terdapat operator pada stack, pop sisanya
    while not isEmpty(s):
        output.append(pop(s))
        
    return "".join(output)

# ==========================================
# Contoh Penggunaan
# ==========================================
ekspresi_infix = "a * b + c / d"
hasil_postfix = infix_ke_postfix(ekspresi_infix)

print(f"Ekspresi Infix   : {ekspresi_infix}")
print(f"Ekspresi Postfix : {hasil_postfix}")
```

#### Code 2

```python
# fungsi menentukan precedence operator
def precedence(op):
    if op == '+' or op == '-':
        return 1
    elif op == '*' or op == '/':
        return 2
    elif op == '^':
        return 3
    else:
        return 0
```

```python
# fungsi mengecek apakah operand
def isOperand(ch):
    return ch.isalnum()   # huruf atau angka
```

```python
# fungsi utama konversi infix ke postfix
def infixToPostfix(infix):
    s = stack()
    output = ""

    for token in infix:

        # jika operand → langsung ke output
        if isOperand(token):
            output += token

        # jika kurung buka → push ke stack
        elif token == '(':
            push(s, token)

        # jika kurung tutup → pop sampai '('
        elif token == ')':
            while not isEmpty(s) and peek(s) != '(':
                output += pop(s)
            pop(s)  # buang '('

        # jika operator
        else:
            while (not isEmpty(s) and precedence(peek(s)) >= precedence(token)):
                output += pop(s)
            push(s, token)

    # pop semua operator sisa
    while not isEmpty(s):
        output += pop(s)

    return output
```

### **Evaluasi Ekspresi Postfix**

Seperti yang sudah dijelaskan sebelumnya, evaluasi untuk ekspressi aritmatika Postfix dilakukan
setelah terdapat dua buah operand sebelum sebuah operator. Misalkan : 87+2*, yang berarti
(8+7)*2=30, oleh karena itu berikut algoritma untuk mengevaluasi ekspressi aritmatika postfix :
1. baca string mulai dari kiri
2. Jika character yang dibaca adalah sebuah operand, maka push operand tersebut.
3. Jika character yang dibaca adalah sebuah operator, maka pop dua buah operand yang
terdapat pada stack, operasikan dua buah operand tersebut dengan operator, dan push hasil
operasinya
4. hasil akhir adalah angka terakhir yang terdapat di dalam stack

```python
def evaluatePost(postStr):
    operandStack=stack()
    operator='+-/*'
    for i in postStr:
        if i not in operator:
            push(operandStack,i)
        else:
            oprnd2=pop(operandStack)
            oprnd1=pop(operandStack)
            if i=='+':
                result=float(oprnd1)+float(oprnd2)
            elif i=='-':
                result=float(oprnd1)-float(oprnd2)
            elif i=='*':
                result=float(oprnd1)*float(oprnd2)
            else:
                result=float(oprnd1)/float(oprnd2)
            push(operandStack,result)
    return(pop(operandStack))
```

### **Evaluasi Ekspresi Prefix**

Berikut algoritma untuk mengevaluasi ekspressi aritmatika postfix :
1. baca string mulai dari kanan
2. Jika bertemu Operand (angka)
3. Jika bertemu Operator:
   - Keluarkan (pop) 2 angka dari stack
   - Aturan Penting Prefix: Angka pertama yang keluar menjadi operand kiri (Operand 1),dan angka kedua yang keluar menjadi operand kanan (Operand 2). (Ini kebalikan dari Postfix).
   - Hitung hasilnya berdasarkan operator (Operand1 Operator Operand2).
   - Masukkan (push) hasil perhitungan tersebut kembali ke stack.
4. Hasil akhir

```python
def hitung(op1,op2,operator):
    if operator == "+":
        return op1 + op2
    if operator == "-":
        return op1 - op2
    if operator == "*":
        return op1 * op2
    if operator == "/":
        return op1 / op2
    if operator == "^":
        return op1 ** op2
    return 0
```

```python
def isAngka(token):
    if token[0] == "-" and size(token) > 1:
        sisaTeks = token[1:]
        return sisaTeks.isdigit()

    else:
        return token.isdigit()
```

```python
def evaluatePrefix(ekspresi):
    s = stack()
    operator = ["+","-","*","/","^"]
    tokenList = ekspresi.split()[::-1]

    for token in tokenList:
        if token in operator:
            if size(s) < 2:
                return "Error: Ekspresi tidak valid (kekurangan angka)!"

            op1 = pop(s)
            op2 = pop(s)

            if token == "/" and op2 == 0:
                return "Error: Pembagian dengan nol!"

            hasil = hitung(op1,op2,token)
            push(s,hasil)

        elif isAngka(token):
            push(s,int(token))

        else:
            return f"Error: Karakter '{token}' tidak dikenali"

    if size(s) == 1:
        return s[0]
    else:
        return "Error: Ekspresi tidak valid (terlalu banyak angka)!"
```

```python
# Infix: (10 * 2) + 15  ==> Hasil: 35
# Prefix: + * 10 2 15
ekspresi_1 = "/ + 7 8 + 3 2"
print(f"Ekspresi Prefix : {ekspresi_1}")
print(f"Hasil Evaluasi  : {evaluatePrefix(ekspresi_1)}\n")
```

```python
# Infix: 100 - (50 / 2) ==> Hasil: 75
# Prefix: - 100 / 50 2
ekspresi_2 = "- 100 / 50 2"
print(f"Ekspresi Prefix : {ekspresi_2}")
print(f"Hasil Evaluasi  : {evaluatePrefix(ekspresi_2)}\n")
```

### **Konversi Postfix ke Infix**

Berikut algoritma untuk konversi ekspressi aritmatik postfix ke infix :
1. Pecah Ekspresi
2. Baca dari Kiri ke Kanan
3. Jika token adalah operand, push token ke dalam stack
4. Jika token adalah Operator
   - Keluarkan (pop) 2 elemen dari stack
   - Elemen pertama yang keluar menjadi operand kanan dan elemen kedua yang keluar menjadi operand kiri
   - Gabungkan keduanya dengan operator di tengah dan tambahkan tanda kurung
   - Masukkan (push) hasil gabungan tersebut kembali ke dalam stack
5. Hasil akhir

```python
def postfixToInfix(ekspresi):
    s = stack()
    operator = ["+","-","*","/","^"]
    tokenList = ekspresi.split()

    for token in tokenList:
        if token in operator:
            if size(s) <2:
                return "Error: Ekspresi tidak valid (kekurangan operand)!"

            kanan = pop(s)
            kiri = pop(s)

            gabungan = f"{kiri} {token} {kanan}"
            push(s,gabungan)

        else:
            push(s,token)

    if size(s) == 1:
        return s[0]
    else:
        return "Error: Ekspresi tidak valid (terlalu banyak angka)!"
```

```python
# Contoh 1: Menggunakan Angka
# Postfix: 10 2 * 15 +
# Infix: ((10 * 2) + 15)
ekspresi_1 = "ABC*+"
print(f"Ekspresi Postfix : {ekspresi_1}")
print(f"Ekspresi Infix   : {postfixToInfix(ekspresi_1)}\n")
```

```python
# Contoh 2: Menggunakan Huruf (Variabel)
# Postfix: A B + C D - *
# Infix: ((A + B) * (C - D))
ekspresi_2 = "A B C * +"
print(f"Ekspresi Postfix : {ekspresi_2}")
print(f"Ekspresi Infix   : {postfixToInfix(ekspresi_2)}")
```

### **Konversi Prefix ke Infix**

Berikut algoritma untuk konversi ekspressi aritmatik prefix ke infix :
1. Pecah dan balik Ekspresi
2. Jika token adalah operand, push token ke dalam stack
3. Jika token adalah Operator
   - Keluarkan (pop) 2 elemen dari stack
   - Elemen pertama yang keluar menjadi operand kanan dan elemen kedua yang keluar menjadi operand kiri
   - Gabungkan keduanya dengan operator di tengah dan tambahkan tanda kurung
   - Masukkan (push) hasil gabungan tersebut kembali ke dalam stack
4. Hasil akhir

```python
def prefixToInfix(ekspresi):
    s = stack()
    operator = ["+","-","*","/","^"]
    tokenList = ekspresi.split()[::-1]

    for token in tokenList:
        if token in operator:
            if size(s) <2:
                return "Error: Ekspresi tidak valid (kekurangan operand)!"

            kanan = pop(s)
            kiri = pop(s)
    
            gabungan = f"{kiri} {token} {kanan}"
            push(s,gabungan)

        else:
            push(s,token)

    if size(s) == 1:
        return s[0]
    else:
        return "Error: Ekspresi tidak valid (terlalu banyak angka)!"
```

```python
# Contoh 1: Menggunakan Angka
# Infix: ((10 * 2) + 15)
# Prefix: + * 10 2 15
ekspresi_1 = "+ * 10 2 15"
print(f"Ekspresi Prefix : {ekspresi_1}")
print(f"Ekspresi Infix  : {prefixToInfix(ekspresi_1)}\n")
```

```python
def balikKata(kalimat):
    s = stack()
    hasil = ""

    for ch in kalimat:
        push(s,ch)

    while not isEmpty(s):
        hapus = pop(s)
        hasil += hapus

    return hasil

# Main Program
kata = "komputer"
print(balikKata(kata))
```

4. Cek keseimbangan tanda kurung menggunakan stack.

```python
def parenthesesCheck(strMath):
  s = stack()
  process = ""
  processError = ""
  matched = True

  count = 0
  countError = 0

  kurung = {")":"(","}":"{","]":"["}
  kurungBuka = kurung.values()
  kurungTutup = kurung.keys()

  for ch in strMath:
    if ch in kurungBuka:
      push(s,ch)
      count += 1
      process += f"{count}. Push stack with {ch} -->{s}\n"
    elif ch in kurungTutup:
      count += 1
      if isEmpty(s):
        matched = False
        countError += 1
        processError += f"{countError}. Kelebihan Kurung Tutup '{ch}' (Stack kosong)\n"
        process += f"{count}. Pop Stack --> Stack Kosong\n"
      else:
        temp = pop(s)
        process += f"{count}. Pop Stack --> {s}\n"

        if kurung[ch] != temp:
          matched = False
          countError += 1
          processError += f"{countError}. Kurung buka '{temp}' tidak cocok dengan kurung tutup {ch}\n"

  if not isEmpty(s):
    matched = False
    for k in s:
      countError += 1
      processError += f"{countError}. Kurung Buka '{k}' belum ditutup\n"

  return matched,process,processError


# Main Program
mathStr = '([4+5]/{9+8]*(3+2]}'
check,proc,err = parenthesesCheck(mathStr)
print(check)
print("Proses Cek : ")
print(proc)
print("Error : ")
print(err)
```

6. Stack dengan batas maksimum (overflow).

```python
# ========================================== #
#         FUNGSI-FUNGSI OPERASI STACK        #
# ========================================== #

def buat_stack():
    # Membuat dan mengembalikan list kosong
    return []

def is_empty(stack):
    return len(stack) == 0

def is_full(stack, kapasitas_maksimal):
    return len(stack) == kapasitas_maksimal

def push(stack, item, kapasitas_maksimal):
    # Cek kondisi OVERFLOW
    if is_full(stack, kapasitas_maksimal):
        print(f"[OVERFLOW] Gagal memasukkan '{item}'. Stack sudah penuh! (Maks: {kapasitas_maksimal})")
        return False
        
    stack.append(item)
    print(f"Push: '{item}' berhasil dimasukkan.")
    return True

def pop(stack):
    # Cek kondisi UNDERFLOW
    if is_empty(stack):
        print("[UNDERFLOW] Gagal pop. Stack kosong!")
        return None
        
    item_popped = stack.pop()
    print(f"Pop : '{item_popped}' dikeluarkan.")
    return item_popped

def peek(stack):
    if is_empty(stack):
        return "Stack kosong!"
    return stack[-1]

def size(stack):
    return len(stack)

# ========================================== #
#             CONTOH PENGGUNAAN              #
# ========================================== #

# 1. Tentukan batas maksimal dan buat stack-nya (berupa list biasa)
batas_maksimal = 3
kardus_buku = buat_stack()

print("--- Memasukkan data ke Stack ---")
# Perhatikan bahwa kita harus "mengirim" variabel kardus_buku ke dalam fungsi
push(kardus_buku, "Buku Matematika", batas_maksimal)
push(kardus_buku, "Buku Fisika", batas_maksimal)
push(kardus_buku, "Buku Kimia", batas_maksimal)

print("\n--- Mencoba memicu OVERFLOW ---")
push(kardus_buku, "Buku Biologi", batas_maksimal) 
push(kardus_buku, "Buku Sejarah", batas_maksimal)

print("\n--- Mengeluarkan data dari Stack ---")
pop(kardus_buku)
pop(kardus_buku)

print("\n--- Cek Ukuran dan Isi ---")
print("Sisa buku di stack:", size(kardus_buku))
print("Buku paling atas  :", peek(kardus_buku))
```

**7.** Hitung jumlah elemen genap dalam stack.

```python
# (Asumsi fungsi buat_stack, is_empty, push, dan pop sudah ada seperti kode sebelumnya)

def hitung_elemen_genap(stack_asli, kapasitas_maksimal):
    stack_sementara = buat_stack()
    jumlah_genap = 0
    
    print("\n--- Memulai Proses Penghitungan Elemen Genap ---")
    
    # Langkah 1-4: Keluarkan, cek, dan pindahkan ke stack sementara
    while not is_empty(stack_asli):
        elemen = pop(stack_asli)
        
        # Memastikan elemen adalah angka dan genap (habis dibagi 2)
        if isinstance(elemen, int) and elemen % 2 == 0:
            jumlah_genap += 1
            
        # Simpan ke stack sementara
        push(stack_sementara, elemen, kapasitas_maksimal)
        
    print(f">> Ditemukan {jumlah_genap} angka genap.")
    print("--- Mengembalikan data ke stack asli ---")
        
    # Langkah 5: Kembalikan semua elemen ke stack asli
    # Karena proses pop dan push dibalik lagi, urutan aslinya akan kembali seperti semula!
    while not is_empty(stack_sementara):
        elemen_kembali = pop(stack_sementara)
        push(stack_asli, elemen_kembali, kapasitas_maksimal)
        
    return jumlah_genap

# ========================================== #
#             CONTOH PENGGUNAAN              #
# ========================================== #

batas_maks = 5
stack_angka = buat_stack()

print("--- Memasukkan Angka ke Stack ---")
push(stack_angka, 15, batas_maks)
push(stack_angka, 8, batas_maks)  # Genap
push(stack_angka, 21, batas_maks)
push(stack_angka, 4, batas_maks)  # Genap
push(stack_angka, 10, batas_maks) # Genap

# Menghitung angka genap
total_genap = hitung_elemen_genap(stack_angka, batas_maks)

print("\n--- Hasil Akhir ---")
print(f"Total elemen genap di dalam stack: {total_genap}")

# Membuktikan bahwa stack asli kembali ke posisi semula (tidak rusak/hilang)
print("Ukuran stack asli sekarang:", size(stack_angka))
print("Elemen paling atas saat ini:", peek(stack_angka))
```
