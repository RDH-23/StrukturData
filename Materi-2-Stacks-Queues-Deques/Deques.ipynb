## **Definisi Deques**

Deque (singkatan dari Double-Ended Queue, biasa dibaca "dek") adalah struktur data linear yang memungkinkan operasi penambahan (insertion) dan penghapusan (deletion) elemen dilakukan dari **kedua ujungnya**, yaitu dari bagian depan (front) maupun bagian belakang (rear).
<br><br>
Berbeda dengan struktur data Queue biasa yang sangat kaku memegang prinsip FIFO (First In First Out) atau Stack yang berprinsip LIFO (Last In First Out), Deque menggabungkan fleksibilitas keduanya.

## **Operasi Deques**

Operas-operasi yang terdapat pada *Deque* antara lain:
- deque (), inisialisasi struktur data deque kosong
- addFront (data), penambahan data baru di ujung front pada deque
- addRear (data), penambahan data baru di ujung rear pada deque
- removeRear (), penghapusan data pada ujung rear
- removeFront (), penghapusan data pada ujung front
- isEmpty(), pengecekan apakah deque dalam keadaan kosong
- size (), informasi jumlah data yang terdapat pada deque

Struktur data *Deque* ini dipresentsikan dengan *List*, dimana:
- indeks awal list, yaitu indeks ke-0, adalah posisi front pada deques
- indeks akhir list, yaitu indeks terakhir, adalah posisi rear pada deques

Oleh karena itu, penambahan data baru pada posisi **front** dapat dilakukan dengan method
insert() pada posisi ke-0. Sedangkan penambahan data baru pada posisi **rear** dapat dilakukan,br dengan method append , yang merupakan penambahan data baru pada list di posisi terakhir.<br><br>
Sedangkan penghapusan data pada posisi **front** dapat dilakukan dengan method pop() pada
posisi ke-0, sedangkan penghapusan data pada posisi **rear** dilakukan dengan method pop() .

## **Code**

Berikut adalah implementasi deques dengan menggunakan List:   

```python
def createDeque():
    d=[]
    return (d)
def addFront(d,data): #front: berada di indeks-0
    d.insert(0,data)
    return(d)
def addRear(d,data): #rear: berada di indeks terakhir
    d.append(data)
    return(d)
def removeRear(d):
    data=d.pop()
    return(data)
def removeFront(d):
    data=d.pop(0)
    return(data)
def isEmpty(d):
    return (d==[])
def size(d):
    return (len(d))
```

# **Implementasi Deques**

Contoh implementasi Deques adalah pengecekan apakah suatu kata merupakan palindrom.
Palindrom merupakan rangkaian kata atau bilangan yang terbaca sama baik dari depan maupun
dari belakang.<br>
Struktur data deques paling tepat digunakan untuk memeriksa apakah suatu kata maupun
bilangan adalah palindrom. Untuk memeriksa apakah suatu kata adalah palindrome, dapat
dilakukan dengan memeriksa apakah ujung yang terletak pada front dan ujung yang terletak pada
rear, merupakan karakter yang sama.

## **Code**

Berikut adalah code untuk pengecekan palindrom dari satu kata

```python
def cekPalindrom(string):
    palindrome=createDeque()
    for huruf in string:
        addRear(palindrome,huruf)
    cek=True
    while size(palindrome)>1:
        a=removeRear(palindrome)
        b=removeFront(palindrome)
        if a==b:
            cek=cek and True
        else:
            cek=cek and False
    return cek
```

```python
print(cekPalindrom('hannah'))
print(cekPalindrom('surabaya'))
print(cekPalindrom('abcdcba'))
print(cekPalindrom('katak'))
print(cekPalindrom('taat'))
print(cekPalindrom('dia'))
```

**Soal 1: Sistem Antrean Pasien (Reguler vs VIP/Darurat)**
Di sebuah klinik, pasien biasa akan mengantre di urutan paling belakang. Namun, jika ada pasien VIP atau keadaan darurat, mereka akan langsung dimasukkan ke urutan paling depan.
Buatlah dua fungsi:

1. tambah_pasien(d, nama, is_vip): Jika is_vip bernilai True, gunakan addFront. Jika False, gunakan addRear.

2. panggil_pasien(d): Panggil pasien dari urutan paling depan menggunakan removeFront.

```python
def tambah_pasien(d, nama, is_vip):
    if is_vip == True:
        addFront(d, nama)
        print(f"[DARURAT/VIP] {nama} dimasukkan ke urutan TERDEPAN.")
    else:
        addRear(d, nama)
        print(f"[REGULER] {nama} mengantre di urutan BELAKANG.")

def panggil_pasien(d):
    if isEmpty(d):
        print("Antrean kosong, dokter bisa istirahat.")
    else:
        pasien = removeFront(d)
        print(f"--> Memanggil pasien ke ruang periksa: {pasien}")

# --- Contoh Penggunaan ---
antrean_klinik = createDeque()

tambah_pasien(antrean_klinik, "Andi", False)
tambah_pasien(antrean_klinik, "Budi", False)
tambah_pasien(antrean_klinik, "Bapak Presiden", True) # VIP datang belakangan
tambah_pasien(antrean_klinik, "Siti", False)

print(f"\nTotal antrean saat ini: {size(antrean_klinik)}\n")

panggil_pasien(antrean_klinik) # Akan memanggil Bapak Presiden duluan
panggil_pasien(antrean_klinik)
panggil_pasien(antrean_klinik)
panggil_pasien(antrean_klinik)
```

**Soal 3: Menyusun Karakter Selang-Seling**
Buatlah fungsi susun_selang_seling(daftar_karakter). Fungsi ini menerima sebuah list berisi huruf-huruf.

- Tugas: Masukkan huruf-huruf tersebut ke dalam Deque secara bergantian: huruf pertama ke depan, huruf kedua ke belakang, huruf ketiga ke depan, huruf keempat ke belakang, dan seterusnya.

- Output: Keluarkan semua isi Deque dari depan ke belakang, lalu gabungkan menjadi satu string utuh dan kembalikan string tersebut.

```python
def susun_selang_seling(daftar_karakter):
    d = createDeque()
    giliran_ke_depan = True # Penanda untuk selang-seling
    
    for huruf in daftar_karakter:
        if giliran_ke_depan == True:
            addFront(d, huruf)
        else:
            addRear(d, huruf)
            
        # Balikkan nilai penanda untuk iterasi berikutnya (True jadi False, False jadi True)
        giliran_ke_depan = not giliran_ke_depan 
        
    # Mengeluarkan semua karakter dari depan untuk melihat hasilnya
    hasil_kata = ""
    while not isEmpty(d):
        hasil_kata += removeFront(d)
        
    return hasil_kata

# --- Contoh Penggunaan ---
huruf_acak = ['A', 'B', 'C', 'D', 'E']
# Ekspektasi proses:
# 1. 'A' ke depan -> [A]
# 2. 'B' ke belakang -> [A, B]
# 3. 'C' ke depan -> [C, A, B]
# 4. 'D' ke belakang -> [C, A, B, D]
# 5. 'E' ke depan -> [E, C, A, B, D]

hasil = susun_selang_seling(huruf_acak)
print(f"Hasil susunan selang-seling: {hasil}")
```
